```python
import os

def verify_environment():
    expected_paths = [
        "ai/schemas",
        "ai/prompts",
        "ai/modules",
        "tests/fixtures",
        "tests/ai_eval",
        "ai/schemas/ai_outputs.py" # الملف الذي أنشأناه للتو
    ]

    print("🔍 جاري فحص بنية مجلدات المشروع...\n")
    
    all_passed = True
    for path in expected_paths:
        if os.path.exists(path):
            print(f"✅ مسار سليم : {path}")
        else:
            print(f"❌ مسار مفقود: {path}")
            all_passed = False

    print("-" * 40)
    if all_passed:
        print("🚀 النتيجة: البنية التحتية صحيحة 100%. نحن جاهزون للخطوة التالية!")
    else:
        print("⚠️ النتيجة: هناك مسارات مفقودة، يرجى مراجعتها وتصحيحها.")

if __name__ == "__main__":
    verify_environment()
```
